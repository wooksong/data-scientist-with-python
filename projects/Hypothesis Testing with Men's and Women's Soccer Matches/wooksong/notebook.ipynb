You're working as a sports journalist at a major online sports media company, specializing in soccer analysis and reporting. You've been watching both men's and women's international soccer matches for a number of years, and your gut instinct tells you that more goals are scored in women's international football matches than men's. This would make an interesting investigative article that your subscribers are bound to love, but you'll need to perform a valid statistical hypothesis test to be sure!

While scoping this project, you acknowledge that the sport has changed a lot over the years, and performances likely vary a lot depending on the tournament, so you decide to limit the data used in the analysis to only official `FIFA World Cup` matches (not including qualifiers) since `2002-01-01`.

You create two datasets containing the results of every official men's and women's international football match since the 19th century, which you scraped from a reliable online source. This data is stored in two CSV files: `women_results.csv` and `men_results.csv`.

The question you are trying to determine the answer to is:

> Are more goals scored in women's international soccer matches than men's?

You assume a **10% significance level**, and use the following null and alternative hypotheses:

$H_0$ : The mean number of goals scored in women's international soccer matches is the same as men's.

$H_A$ : The mean number of goals scored in women's international soccer matches is greater than men's.

```python
# Start your code here!
import pandas as pd
```

```python
# you that more goals are scored in women's international football matches than men's.
# H0: score_women = score_men
# H1: score_women > score_men

df_men_results = pd.read_csv('men_results.csv')
df_women_results = pd.read_csv('women_results.csv')
```

```python
df_men_results.head()
```

```python
wc_men_results = df_men_results[df_men_results['tournament'] == 'FIFA World Cup']
wc_women_results = df_women_results[df_women_results['tournament'] == 'FIFA World Cup']
```

```python
wc_men_results = wc_men_results[wc_men_results['date'] > '2002-01-01']
wc_women_results = wc_women_results[wc_women_results['date'] > '2002-01-01']
```

```python
wc_men_scores = wc_men_results['home_score'] + wc_men_results['away_score']
wc_women_scores = wc_women_results['home_score'] + wc_women_results['away_score']

ndarray_m = wc_men_scores.to_numpy()
ndarray_w = wc_women_scores.to_numpy()
```

```python
from scipy.stats import shapiro

# Shapiro-Wilk Test
stat_sw_m, p_sw_m = shapiro(ndarray_m)
stat_sw_w, p_sw_w = shapiro(ndarray_w)

print("Shapiro-Wilk Test")
print(f"  Men's Score: Stat={stat_sw_m}, p={p_sw_m}, it’s {'not' if p_sw_m < 0.05 else ''} normally distributed.")
print(f"  Women's Score: Stat={stat_sw_w}, p={p_sw_w}, it’s {'not' if p_sw_w < 0.05 else ''} normally distributed.")
```

```python
from scipy.stats import kstest

# K-S Test
stat_ks_m, p_ks_m = kstest(ndarray_m, 'norm', args=(ndarray_m.mean(), ndarray_m.std()))
stat_ks_w, p_ks_w = kstest(ndarray_w, 'norm', args=(ndarray_w.mean(), ndarray_w.std()))
print("K-S Test")
print(f"  Men's Score: Stat={stat_ks_m}, p={p_ks_m}, it’s {'not' if p_ks_m < 0.05 else ''} normally distributed.")
print(f"  Women's Score: Stat={stat_ks_w}, p={p_ks_w}, it’s {'not' if p_ks_w < 0.05 else ''} normally distributed.")
```

```python
from scipy.stats import mannwhitneyu

result_mw = mannwhitneyu(x=ndarray_m, y=ndarray_w, alternative='less')
p_val = result_mw.pvalue
result_dict = {"p_val": p_val, "result": "reject" if p_val <= 0.1 else "fail to reject"}
```
